# **Explore the folder structure**

```python
import os

dataset_path = "plant_dataset"

for root, dirs, files_ in os.walk(dataset_path):
    level = root.replace(dataset_path, "").count(os.sep)
    indent = " " * 2 * level
    print(f"{indent}{os.path.basename(root)}/")
    if level < 2:
        for f in files_[:3]:
            print(f"{indent}  {f}")
```

# Check for empty class folders

```python
for class_name in os.listdir(dataset_path):
    class_path = os.path.join(dataset_path, class_name)
    if os.path.isdir(class_path):
        num_files = len(os.listdir(class_path))
        if num_files == 0:
            print(f"⚠️ EMPTY folder: {class_name}")
        else:
            print(f"{class_name}: {num_files} images")
```

# **Find corrupted / unreadable images**

```python
from PIL import Image
import os

corrupted_files = []

for root, dirs, files_ in os.walk(dataset_path):
    for file in files_:
        file_path = os.path.join(root, file)
        try:
            img = Image.open(file_path)
            img.verify()  # checks the file is a valid image
        except Exception as e:
            corrupted_files.append(file_path)

print(f"Found {len(corrupted_files)} corrupted/unreadable images")
for f in corrupted_files:
    print(f)
```

# **Find zero-byte (empty) files**

```python
zero_byte_files = []

for root, dirs, files_ in os.walk(dataset_path):
    for file in files_:
        file_path = os.path.join(root, file)
        if os.path.getsize(file_path) == 0:
            zero_byte_files.append(file_path)

print(f"Found {len(zero_byte_files)} zero-byte files")
```

# **Remove the bad files (corrupted + zero-byte)**

```python
files_to_remove = set(corrupted_files) | set(zero_byte_files)

for f in files_to_remove:
    os.remove(f)
    print(f"Removed: {f}")

print(f"\nTotal removed: {len(files_to_remove)}")
```

# **Find and remove exact duplicate images**

```python
import hashlib

def file_hash(path):
    with open(path, "rb") as f:
        return hashlib.md5(f.read()).hexdigest()

seen_hashes = {}
duplicates = []

for root, dirs, files_ in os.walk(dataset_path):
    for file in files_:
        file_path = os.path.join(root, file)
        try:
            h = file_hash(file_path)
            if h in seen_hashes:
                duplicates.append(file_path)
            else:
                seen_hashes[h] = file_path
        except Exception:
            pass

print(f"Found {len(duplicates)} duplicate images")

for f in duplicates:
    os.remove(f)
print(f"Removed {len(duplicates)} duplicates")
```

# **Final class balance check (after cleaning)**

```python
import pandas as pd

summary = []
for class_name in os.listdir(dataset_path):
    class_path = os.path.join(dataset_path, class_name)
    if os.path.isdir(class_path):
        summary.append({"class": class_name, "num_images": len(os.listdir(class_path))})

df = pd.DataFrame(summary).sort_values("num_images", ascending=False)
print(df)
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(10, 6))
sns.histplot(df["num_images"], bins=15, kde=True, color="steelblue")
plt.axvline(df["num_images"].mean(), color="red", linestyle="--", label=f"Mean = {df['num_images'].mean():.0f}")
plt.xlabel("Number of Images per Class")
plt.ylabel("Number of Classes")
plt.title("Distribution of Images per Disease Class (After Cleaning)")
plt.legend()
plt.tight_layout()
plt.show()
```
